```python
# ==========================================
# 2. DATA GOVERNANCE: Column-Level Security (Dynamiczne maskowanie)
# ==========================================
print("⏳ Tworzenie bezpiecznego widoku (Dynamic Data Masking)...")

spark.sql("""
CREATE OR REPLACE VIEW dbw_showcase.default.secure_vw_payroll_gold AS
SELECT 
    department_name,
    city,
    total_transactions,
    CASE 
        -- Wersja demonstracyjna na potrzeby portfolio
        WHEN current_user() = 'dyrektor.hr@twojafirma.com' THEN CAST(total_hours AS STRING)
        ELSE 'ZAMASKOWANE'
    END AS masked_total_hours
FROM dbw_showcase.default.payroll_gold
""")

print("✅ Bezpieczny widok z maskowaniem na poziomie kolumny utworzony!")

# ==========================================
# 3. TEST BEZPIECZEŃSTWA (Ścieżka negatywna)
# ==========================================
print("\n🔍 Wyświetlam wynik z bezpiecznego widoku dla Ciebie (powinieneś zobaczyć maskowanie!):")
display(spark.sql("SELECT * FROM dbw_showcase.default.secure_vw_payroll_gold"))
```
